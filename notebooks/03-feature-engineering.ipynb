# 03 — Feature engineering and MCDA scoring

300 m square cells, OSM counts and areal-weighted population feed an explainable score.

Positive contributions: demand 45%, access 25%, population 20%, normalized by their own total. Restaurant saturation subtracts up to 3 points; nested café competition subtracts up to 7 points. These are scenario weights, not profit-trained coefficients. Missing penalties make the score provisional. New school counts require explicit ISCED 2/3 tags; the committed legacy dataset uses generic school counts.
