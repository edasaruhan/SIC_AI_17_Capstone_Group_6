```python
import geopandas as gpd
import matplotlib.pyplot as plt
from matplotlib.patches import Patch
from matplotlib.lines import Line2D

raw = ROOT / "data/raw"
boundary = gpd.read_file(raw / "cankaya_boundary.geojson")
layers = {
    "park": (gpd.read_file(raw / "cankaya_parks.geojson"), "#27ae60", 6),
    "shop": (gpd.read_file(raw / "cankaya_shops.geojson"), "#f1c40f", 3),
    "school": (gpd.read_file(raw / "cankaya_schools.geojson"), "#2980b9", 8),
    "hospital": (gpd.read_file(raw / "cankaya_hospitals.geojson"), "#db7093", 10),
    "university": (gpd.read_file(raw / "cankaya_universities.geojson"), "#8e44ad", 14),
    "bus": (gpd.read_file(raw / "cankaya_bus_stops.geojson"), "#16a085", 4),
    "metro": (gpd.read_file(raw / "cankaya_metro.geojson"), "#2c3e50", 12),
    "restaurant": (gpd.read_file(raw / "cankaya_restaurants.geojson"), "#e67e22", 6),
    "cafe": (gpd.read_file(raw / "cankaya_cafes.geojson"), "#dc2626", 8),
}

fig, ax = plt.subplots(figsize=(9, 9))
boundary.plot(ax=ax, facecolor="#eef2ff", edgecolor="#1e3a8a", linewidth=1.1)
for _name, (gdf, color, size) in layers.items():
    if not gdf.empty:
        gdf.plot(ax=ax, color=color, markersize=size, alpha=0.75)
ax.set_title("Cankaya OSM backbone (venues, demand, transit)")
ax.set_axis_off()
handles = [
    Patch(facecolor="#eef2ff", edgecolor="#1e3a8a", label="Boundary"),
    Line2D([0], [0], marker="o", color="w", markerfacecolor="#dc2626", markersize=8, label="Cafe"),
    Line2D([0], [0], marker="o", color="w", markerfacecolor="#e67e22", markersize=8, label="Restaurant"),
    Line2D([0], [0], marker="o", color="w", markerfacecolor="#8e44ad", markersize=8, label="University"),
    Line2D([0], [0], marker="o", color="w", markerfacecolor="#2980b9", markersize=8, label="School"),
    Line2D([0], [0], marker="o", color="w", markerfacecolor="#db7093", markersize=8, label="Hospital"),
    Line2D([0], [0], marker="o", color="w", markerfacecolor="#27ae60", markersize=8, label="Park"),
    Line2D([0], [0], marker="o", color="w", markerfacecolor="#f1c40f", markersize=8, label="Shop"),
    Line2D([0], [0], marker="o", color="w", markerfacecolor="#16a085", markersize=8, label="Bus stop"),
    Line2D([0], [0], marker="o", color="w", markerfacecolor="#2c3e50", markersize=8, label="Metro"),
]
ax.legend(handles=handles, loc="lower left", fontsize=8, framealpha=0.9)
fig.tight_layout()
(ROOT / "images").mkdir(exist_ok=True)
fig.savefig(ROOT / "images/suitability-map.png", dpi=140, bbox_inches="tight")
plt.show()
```
