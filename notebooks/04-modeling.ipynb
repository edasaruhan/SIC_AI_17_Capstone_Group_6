The Random Forest is implemented in `src/cafe_similarity.py` and used by the app.

- Target: at least one mapped café within 500 m (`cafes_500m >= 1`), not inside the cell alone.
- Café counts are excluded from features. POI diversity is recomputed without cafés, including for legacy data.
- Whole mahalle groups are held out; border-adjacent cells can still share spatial signals.
- Missing-value medians are fitted on the training fold only.
- Displayed scores are held-out probabilities × 100. The full-data fit is used only for feature importances.
- Run metadata and spatial ROC-AUC are documented in `docs/model_validation.md`. The model does not predict revenue, profit or commercial success.
