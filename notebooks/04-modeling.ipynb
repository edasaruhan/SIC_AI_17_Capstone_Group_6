# 04 — Optional similarity models (not profit)

Do **not** label cells with cafés as good locations and empty cells as bad ones. That only answers “where do cafés already exist?”.

If you train logistic regression or Random Forest later:

- Target: `1` if the cell has at least one mapped café, else `0`
- Name the output **existing café location similarity score**
- Hold out whole mahalle polygons for testing (adjacent 300 m cells leak if you split at random)
- Report precision, recall, F1, ROC-AUC against a majority-class baseline
- Never call the probability “chance a café would be profitable”

The capstone score people should trust is still the Week 3 MCDA map (`src/scoring.py`), not this classifier.
